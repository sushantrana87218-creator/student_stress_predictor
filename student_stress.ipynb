<!-- handling missing values -->

```python
data['Student_Type']=data['Student_Type'].fillna(data['Student_Type'].mode()[0])

data['Sleep_Hours']=data['Sleep_Hours'].fillna(data['Sleep_Hours'].median())
data['Study_Hours']=data['Study_Hours'].fillna(data['Study_Hours'].median())
data['Social_Media_Hours']=data['Social_Media_Hours'].fillna(data['Social_Media_Hours'].median())
data['Attendance']=data['Attendance'].fillna(data['Attendance'].median())
data['Exam_Pressure']=data['Exam_Pressure'].fillna(data['Exam_Pressure'].median())
data['Family_Support']=data['Family_Support'].fillna(data['Family_Support'].median())
data['Month']=data['Month'].fillna(data['Month'].median())

data.isnull().sum()
```

```python
data['Student_Type'].unique()
```

```python
data['Student_Type']=data['Student_Type'].map({'school':0,'college':1,'working_student':2})
```

```python
x=data.drop(['Stress_Level'],axis=1)
y=data['Stress_Level']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size=0.20,random_state=42,stratify=y
)
x_test
```

```python
x_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from xgboost import XGBClassifier
```

```python
lr = LogisticRegression(max_iter=1000)
lr.fit(x_train, y_train)

rf = RandomForestClassifier(n_estimators=200, random_state=42)
rf.fit(x_train, y_train)

gb = GradientBoostingClassifier(random_state=42)
gb.fit(x_train, y_train)

xg = XGBClassifier(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    eval_metric='logloss'
)
xg.fit(x_train, y_train)
```

```python
y_pred1=lr.predict(x_test)
y_pred2=rf.predict(x_test)
y_pred3=gb.predict(x_test)
y_pred4=xg.predict(x_test)
```

```python
x_test.head()
```

```python
y_test
```

```python
y_pred1
```

```python
from sklearn import metrics
```

```python
score1=metrics.accuracy_score(y_test,y_pred1)
score2=metrics.accuracy_score(y_test,y_pred2)
score3=metrics.accuracy_score(y_test,y_pred3)
score4=metrics.accuracy_score(y_test,y_pred4)

f1_score1=metrics.f1_score(y_test,y_pred1)
f1_score2=metrics.f1_score(y_test,y_pred2)
f1_score3=metrics.f1_score(y_test,y_pred3)
f1_score4=metrics.f1_score(y_test,y_pred4)
```

```python
final_data= pd.DataFrame({
    'Models':['LR','RF','GBC','XGB'],
    'ACCURACY':[score1,score2,score3,score4],
    'F1_SCORE':[f1_score1,f1_score2,f1_score3,f1_score4]
})
```

```python
final_data
```

```python
ax = sns.barplot(data=final_data, x="Models", y="ACCURACY", color="steelblue")
ax.set_title("Model Performance (Accuracy)")
ax.set_xlabel("Model")
ax.set_ylabel("Accuracy")
ax.set_ylim(0, 1)
ax.bar_label(ax.containers[0], fmt="%.3f", padding=3)
plt.tight_layout()
plt.show()
```

```python
print("XGBoost Classification Report")
print(metrics.classification_report(y_test, y_pred4))

print("XGBoost Confusion Matrix")
print(metrics.confusion_matrix(y_test, y_pred4))
```
